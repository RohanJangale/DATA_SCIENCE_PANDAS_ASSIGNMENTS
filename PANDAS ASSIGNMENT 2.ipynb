7. Creating and Modifying Columns

```python
#Question 29
#Create a new column named AgeGroup. Store Above 40 when Age is greater than 40; otherwise store 40 or Below. Use apply() with a lambda function.
df['AgeGroup']=df['Age'].apply(lambda x: 'Above 40' if x > 40 else ' 40 or Below ')
df
```

```python
#Question 30
#Display the first 10 rows after creating the AgeGroup column.
df.head(10)
```

```python
#Question 31
#Count the values in the newly created AgeGroup column using value_counts().
df['AgeGroup'].value_counts()
```

```python
#Question 32
#Remove the AgeGroup column using drop().
df.drop(columns=['AgeGroup'], inplace=True)
```

```python
#Question 33
#Rename the column BloodPrs to BloodPressure.
df.rename(columns={'Bloodprs':'BloodPressure'},inplace=True)
df
```
